```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("skin clinic campaign.csv")

df.head()
```

```python
print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
def response_rate_table(data, group_column):
    temp = data.copy()

    temp["Responded"] = (
        temp["Response_to_Campaign"]
        .str.strip()
        .str.lower()
        .eq("yes")
    )

    result = (
        temp.groupby(group_column, observed=False)
        .agg(
            Total_Customers=("CustID", "size"),
            Responded=("Responded", "sum"),
            Response_Rate=("Responded", "mean")
        )
        .reset_index()
    )

    result["Response Rate (%)"] = (
        result["Response_Rate"] * 100
    ).round(2)

    result = result.drop(columns="Response_Rate")

    return result
```

```python
gender_analysis = response_rate_table(df, "Gender")

gender_analysis
```

```python
age_analysis = response_rate_table(df, "AgeGroup")

age_analysis
```

```python
purchase_analysis = response_rate_table(
    df,
    "Purchase_Last_Quarter"
)

purchase_analysis
```

```python
df["Product_Usage"] = pd.cut(
    df["Unique_Products_Purchased"],
    bins=[0, 4, 8, np.inf],
    labels=["1-4", "5-8", ">8"]
)

df[[
    "Unique_Products_Purchased",
    "Product_Usage"
]].head()
```

```python
product_analysis = response_rate_table(
    df,
    "Product_Usage"
)

product_analysis
```

```python
print("GENDER VS CAMPAIGN RESPONSE")
display(gender_analysis)

print("AGE GROUP VS CAMPAIGN RESPONSE")
display(age_analysis)

print("PURCHASE IN LAST QUARTER VS CAMPAIGN RESPONSE")
display(purchase_analysis)

print("PRODUCT USAGE VS CAMPAIGN RESPONSE")
display(product_analysis)
```
