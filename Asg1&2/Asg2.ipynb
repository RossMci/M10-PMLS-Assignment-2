```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("skin clinic campaign.csv")

df.head()
```

```python
print("Number of rows:", len(df))
print("Columns:")
print(df.columns.tolist())
```

```python
def response_rate_table(data, group_column):

    temp = data.copy()

    # Convert Yes/No campaign response into True/False
    temp["Responded"] = (
        temp["Response_to_Campaign"]
        .astype(str)
        .str.strip()
        .str.lower()
        .eq("yes")
    )

    # Group customers
    result = (
        temp.groupby(group_column, observed=False)
        .agg(
            Total_Customers=("CustID", "size"),
            Responded=("Responded", "sum"),
            Response_Rate=("Responded", "mean")
        )
        .reset_index()
    )

    # Convert response rate to percentage
    result["Response Rate (%)"] = (
        result["Response_Rate"] * 100
    ).round(2)

    # Remove temporary column
    result = result.drop(
        columns=["Response_Rate"]
    )

    return result
```

```python
gender_analysis = response_rate_table(
    df,
    "Gender"
)

gender_analysis
```

```python
age_analysis = response_rate_table(
    df,
    "AgeGroup"
)

age_analysis
```

```python
purchase_analysis = response_rate_table(
    df,
    "Purchase_Last_Quarter"
)

purchase_analysis
```

```python
df["Product_Usage"] = pd.cut(
    df["Unique_Products_Purchased"],
    bins=[0, 4, 8, np.inf],
    labels=["1-4", "5-8", ">8"]
)
```

```python
df[
    [
        "Unique_Products_Purchased",
        "Product_Usage"
    ]
].head(10)
```

```python
df[
    [
        "Unique_Products_Purchased",
        "Product_Usage"
    ]
].head(10)
```

```python
product_analysis = response_rate_table(
    df,
    "Product_Usage"
)

product_analysis
```

```python
output_file = "M10_PMLS_Assignment_2.xlsx"

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    gender_analysis.to_excel(
        writer,
        sheet_name="Gender",
        index=False
    )

    age_analysis.to_excel(
        writer,
        sheet_name="Age Group",
        index=False
    )

    purchase_analysis.to_excel(
        writer,
        sheet_name="Last Quarter",
        index=False
    )

    product_analysis.to_excel(
        writer,
        sheet_name="Product Usage",
        index=False
    )

print("Excel file created successfully!")
```
